```python
import pandas as pd
```

```python
good_quad = pd.read_csv("goodQuadgram.csv")
good_tri = pd.read_csv("goodTriGrams.csv")
good_bi = pd.read_csv("goodBigram.csv")
good_uni = pd.read_csv(r"D:\ML Internship\test 2\ngrams\unigram_sorted.csv")
```

```python
good_bi = good_bi.loc[good_bi["documents"]>2]
good_uni = good_uni.loc[good_uni["documents"]>2]
```

```python
def extractGrams(gramNo,targetGramNo,data):
    ans = []
    i = 0
    while(i<=(gramNo-targetGramNo)):
        st = ""
        for j in range(targetGramNo):
            st += data[j+i] + " "
        ans.append(st[:-1])
        i += 1
    return ans
```

```python
def remove_duplicate(data):
    seen = set()
    result = []
    for item in data:
        if item[0] not in seen:
            seen.add(item[0])
            result.append(item)
    return result
```

## Extracting grams from Qaudgrams

```python
tri_terms = []
bi_terms = []
uni_terms = []

for i,term in enumerate(good_quad["4-gram"]):
    term_li = term.split(" ")

    tri_li = extractGrams(4,3,term_li)
    for tri in tri_li:
        tri_terms.append((tri,i))

    bi_li = extractGrams(4,2,term_li)
    for bi in bi_li:
        bi_terms.append((bi,i))

    uni_li = extractGrams(4,1,term_li)
    for uni in uni_li:
        uni_terms.append((uni,i))
```

```python
tri_terms = remove_duplicate(tri_terms)
bi_terms = remove_duplicate(bi_terms)
uni_terms = remove_duplicate(uni_terms)
```

```python
for i,tri in enumerate(good_tri["3-gram"]):
    for term in tri_terms:
        if tri==term[0]:
            value = good_tri["total freq"][i] - good_quad["total freq"][term[1]]
            good_tri.loc[i,'total freq'] = value
```

```python
for i, bi in enumerate(good_bi["2-gram"]):
    for term in bi_terms:
        if bi == term[0]:
            value = good_bi["total freq"].iloc[i] - good_quad["total freq"].iloc[term[1]]
            good_bi.loc[i, "total freq"] = value
```

```python
for i, uni in enumerate(good_uni["1-gram"]):
    for term in uni_terms:
        if uni == term[0]:
            value = good_uni["total freq"].iloc[i] - good_quad["total freq"].iloc[term[1]]
            good_uni.loc[i, "total freq"] = value
```
