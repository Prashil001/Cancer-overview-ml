```python
from collections import Counter
from nltk.util import ngrams
import glob
import os
```

```python
def get_ngram_frequencies(folder_path, n):
    ngram_counts = Counter()
    for file in glob.glob(os.path.join(folder_path, "*.txt")):
        with open(file, 'r', encoding='utf-8') as f:
            words = f.read().split()
            ngram_counts.update(ngrams(words, n))
    return ngram_counts
```

```python
output_folder = r"E:\output of cancer crawled"

bigram_counts = get_ngram_frequencies(output_folder, 2)
bigrams = []
for ngram, count in bigram_counts.most_common(100):
    text = f"{ngram[0]} {ngram[1]}"
    bigrams.append(text)

bigrams
```

```python

trigram_counts = get_ngram_frequencies(output_folder, 3)
trigrams = []
for ngram, count in trigram_counts.most_common(100):
    text = f"{ngram[0]} {ngram[1]} {ngram[2]}"
    trigrams.append(text)

trigrams
```

```python
def remove_ngrams_from_text(text, bigrams, trigrams):
    words = text
    i = 0
    new_words = []

    while i < len(words):
        tri = " ".join(words[i:i+3])
        bi = " ".join(words[i:i+2])

        if i + 2 < len(words) and tri in trigrams:
            i += 3  
        elif i + 1 < len(words) and bi in bigrams:
            i += 2  
        else:
            new_words.append(words[i])
            i += 1
    return new_words
```

```python
old_text = []
for file in glob.glob(os.path.join(output_folder, "*.txt")):
    with open(file, 'r', encoding='utf-8') as f:
        words = f.read().split()
        for word in words:
            old_text.append(word)



        
len(old_text)
```

```python
corpus_updated = remove_ngrams_from_text(old_text,bigrams,trigrams)
```

```python
final_output_folder = r"E:\Cancer-overview-ml"
final_output_file = os.path.join(final_output_folder,"final_output.txt")
with open(final_output_file,'a') as f:
    for words in corpus_updated:
        f.write(f"{words} ")
    
```
