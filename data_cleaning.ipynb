```python
custom_stop_words = [
    'nature', 'article', 'journal', 'pubmed', 'google', 'scholar', 'patient', 'case', 'control', 
    'study', 'result', 'author', 'data', 'p', 'ci', 'r', 'g', 'c', 'n', 'et', 'al', 'fig'
]
```

```python
for word in custom_stop_words:
    nlp.Defaults.stop_words.add(word)
    nlp.vocab[word].is_stop = True
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'http\S+|www.\S+', '', text) 
    text = re.sub(r'\S+@\S+', '', text)
    doc = nlp(text)
    
    cleaned_tokens = []
  
    allowed_pos = ['NOUN', 'PROPN', 'ADJ'] 
    entities_to_remove = {'PERSON', 'ORG', 'GPE'}
    
    for token in doc:
        if token.ent_type_ in entities_to_remove:
            continue 
        if (
            not token.is_stop and
            not token.is_punct and
            not token.like_num and
            token.pos_ in allowed_pos and
            len(token.lemma_) > 2 
        ):
            cleaned_tokens.append(token.lemma_)
            
    return cleaned_tokens
```

```python
source_directory = r"D:\projects\output_text_files"
destination_directory = r"D:\projects\cleaned_text_output"

os.makedirs(destination_directory, exist_ok=True)

for filepath in glob.glob(os.path.join(source_directory, "*.txt")):
    
    with open(filepath, 'r', encoding='utf-8') as infile:
        

        cleaned_tokens = clean_text(infile.read())
        
        cleaned_text = " ".join(cleaned_tokens)
        
        filename = os.path.basename(filepath)
        
        destination_path = os.path.join(destination_directory, filename)
        
        with open(destination_path, 'w', encoding='utf-8') as f:
            f.write(cleaned_text)
            
    print(f"Cleaned '{filepath}' and saved to '{destination_path}'")

print("\nAll files have been cleaned and saved to the new directory.")
```
