```python
import os
import glob
import csv
from collections import Counter
import nltk
import pandas as pd
```

```python
cleaned_dir = r"D:\projects\cleaned_text_output"
output_dir = r"D:\projects\ngrams"
os.makedirs(output_dir, exist_ok=True)
```

```python
unigram_df = pd.DataFrame(columns=["unigram","documents","total freq"])
bigram_df = pd.DataFrame(columns=["bigram","documents","total freq"])
trigram_df = pd.DataFrame(columns=["trigram","documents","total freq"])
quadgram_df = pd.DataFrame(columns=["quadgram","documents","total freq"])
```

```python
all_tokens = []
for filepath in glob.glob(os.path.join(cleaned_dir, "*.txt")):
    with open(filepath, 'r', encoding='utf-8') as f:
        text = f.read()
        tokens = text.split()  
        all_tokens.extend(tokens)
```

```python

all_unigram = list(nltk.ngrams(all_tokens,1))
all_unigram_count = Counter(all_unigram)
all_unigram_set = set(all_unigram)

all_bigram = list(nltk.ngrams(all_tokens,2))
all_bigram_count = Counter(all_bigram)
all_bigram_set = set(all_bigram)

all_trigram = list(nltk.ngrams(all_tokens,3))
all_trigram_count = Counter(all_trigram)
all_trigram_set = set(all_trigram)

all_quadgram = list(nltk.ngrams(all_tokens,4))
all_quadgram_count = Counter(all_quadgram)
all_quadgram_set = set(all_quadgram)
```

```python
unigram_tem = pd.DataFrame({"unique unigrams": list(all_unigram_set), "docs": [0]*len(all_unigram_set), "total freq": [0]*len(all_unigram_set)})

bigram_tem = pd.DataFrame({"unique bigrams": list(all_bigram_set), "docs": [0]*len(all_bigram_set), "total freq": [0]*len(all_bigram_set)})

trigram_tem = pd.DataFrame({"unique trigrams": list(all_trigram_set), "docs": [0]*len(all_trigram_set), "total freq": [0]*len(all_trigram_set)})

quadgram_tem = pd.DataFrame({"unique quadgrams": list(all_quadgram_set), "docs": [0]*len(all_quadgram_set), "total freq": [0]*len(all_quadgram_set)})


```

```python


for idx, filepath in enumerate(glob.glob(os.path.join(cleaned_dir, "*.txt"))):
    with open(filepath, 'r', encoding='utf-8') as f:
        text = f.read()
        tokens = text.split()

        unigram = list(nltk.ngrams(tokens, 1))
        unigram_count = Counter(unigram)
        unigram_set = set(unigram)

        for gram in unigram_set:
            if gram in all_unigram_set:
                unigram_tem.loc[unigram_tem['unique unigrams'] == gram, 'docs'] += 1
                unigram_tem.loc[unigram_tem['unique unigrams'] == gram, 'total freq'] = all_unigram_count[gram]


        bigram = list(nltk.ngrams(tokens, 2))
        bigram_count = Counter(bigram)
        bigram_set = set(bigram)

        for gram in bigram_set:
            if gram in all_bigram_set:
                bigram_tem.loc[bigram_tem['unique bigrams'] == gram, 'docs'] += 1
                bigram_tem.loc[bigram_tem['unique bigrams'] == gram, 'total freq'] = all_bigram_count[gram]


        trigram = list(nltk.ngrams(tokens, 3))
        trigram_count = Counter(trigram)
        trigram_set = set(trigram)

        for gram in trigram_set:
            if gram in all_trigram_set:
                trigram_tem.loc[trigram_tem['unique trigrams'] == gram, 'docs'] += 1
                trigram_tem.loc[trigram_tem['unique trigrams'] == gram, 'total freq'] = all_trigram_count[gram]

        quadgram = list(nltk.ngrams(tokens, 4))
        quadgram_count = Counter(quadgram)
        quadgram_set = set(quadgram)

        for gram in quadgram_set:
            if gram in all_quadgram_set:
                quadgram_tem.loc[quadgram_tem['unique quadgrams'] == gram, 'docs'] += 1
                quadgram_tem.loc[quadgram_tem['unique quadgrams'] == gram, 'total freq'] = all_quadgram_count[gram]


```

```python

unigram_tem = unigram_tem.sort_values(by='total freq', ascending=False)
unigram_tem.to_csv(os.path.join(output_dir, "unigram_tem_sorted.csv"), index=False)

bigram_tem = bigram_tem.sort_values(by='total freq', ascending=False)
bigram_tem.to_csv(os.path.join(output_dir, "bigram_tem_sorted.csv"), index=False)

trigram_tem = trigram_tem.sort_values(by='total freq', ascending=False)
trigram_tem.to_csv(os.path.join(output_dir, "trigram_tem_sorted.csv"), index=False)

quadgram_tem = quadgram_tem.sort_values(by='total freq', ascending=False)
quadgram_tem.to_csv(os.path.join(output_dir, "quadgram_tem_sorted.csv"), index=False)
```
