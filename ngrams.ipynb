```python
import os
import glob
import csv
from collections import Counter
import nltk
```

```python
cleaned_dir = r"E:\cancer cleaned text"
output_dir = r"E:\analysis_results\ngrams"
os.makedirs(output_dir, exist_ok=True)
```

```python
TOP_N = 1000
```

```python
all_tokens = []
for filepath in glob.glob(os.path.join(cleaned_dir, "*.txt")):
    with open(filepath, 'r', encoding='utf-8') as f:
        text = f.read()
        tokens = text.split()  
        all_tokens.extend(tokens)
```

```python
bigrams = list(nltk.bigrams(all_tokens))
trigrams = list(nltk.trigrams(all_tokens))
```

```python
bigram_counts = Counter(bigrams)
trigram_counts = Counter(trigrams)
```

```python
bigram_csv = os.path.join(output_dir, "top_bigrams.csv")
with open(bigram_csv, 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["bigram", "count"])
    for (w1, w2), cnt in bigram_counts.most_common(TOP_N):
        writer.writerow([f"{w1} {w2}", cnt])
```

```python
trigram_csv = os.path.join(output_dir, "top_trigrams.csv")
with open(trigram_csv, 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(["trigram", "count"])
    for (w1, w2, w3), cnt in trigram_counts.most_common(TOP_N):
        writer.writerow([f"{w1} {w2} {w3}", cnt])
```

```python
print(f"Top {TOP_N} bigrams saved to: {bigram_csv}")
print(f"Top {TOP_N} trigrams saved to: {trigram_csv}")
```
