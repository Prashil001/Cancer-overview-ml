```python
import pandas as pd
```

```python
unigram = pd.read_csv(r"D:\ML Internship\Test N grams\unigram_tem_sorted.csv")
bigram = pd.read_csv(r"D:\ML Internship\Test N grams\bigram_tem_sorted.csv")
trigram = pd.read_csv(r"D:\ML Internship\Test N grams\trigram_tem_sorted.csv")
quadgram = pd.read_csv(r"D:\ML Internship\Test N grams\quadgram_tem_sorted.csv")
```

```python
bi_quan = []
uni_quan = []
tri_quan = []
quad_quan = []
```

```python
for i in range(1,101):
    q = i/100
    bi_row = {"percentile":i, "documents":bigram['documents'].quantile(q), "total_freq": bigram['total freq'].quantile(q), "avg":bigram["avg"].quantile(q)}
    uni_row = {"percentile":i, "documents":unigram['documents'].quantile(q), "total_freq": unigram['total freq'].quantile(q), "avg":unigram["avg"].quantile(q)}
    tri_row = {"percentile":i, "documents":trigram['documents'].quantile(q), "total_freq": trigram['total freq'].quantile(q), "avg":trigram["avg"].quantile(q)}
    quad_row = {"percentile":i, "documents":quadgram['documents'].quantile(q), "total_freq": quadgram['total freq'].quantile(q), "avg":quadgram["avg"].quantile(q)}
    bi_quan.append(bi_row)
    uni_quan.append(uni_row)
    tri_quan.append(tri_row)
    quad_quan.append(quad_row)
```

```python
bi_quantiled_df = pd.DataFrame(bi_quan)
uni_quantiled_df = pd.DataFrame(uni_quan)
tri_quantiled_df = pd.DataFrame(tri_quan)
quad_quantiled_df = pd.DataFrame(quad_quan)
```

```python
bi_quantiled_df.to_csv(r"D:\ML Internship\Test N grams\bigram_quantiled.csv")
uni_quantiled_df.to_csv(r"D:\ML Internship\Test N grams\unigram_quantiled.csv")
tri_quantiled_df.to_csv(r"D:\ML Internship\Test N grams\trigram_quantiled.csv")
quad_quantiled_df.to_csv(r"D:\ML Internship\Test N grams\quadgram_quantiled.csv")
```
