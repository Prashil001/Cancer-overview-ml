```python
import pandas as pd
```

```python
df = pd.read_csv(r"D:\ML Internship\Test N grams\bigram_tem_sorted.csv")
```

```python
df
```

```python
quantiles = []

for i in range(1,101):
    q = i/100
    row = {"percentile":i, "documents":df['documents'].quantile(q), "total_freq": df['total freq'].quantile(q), "avg":df["avg"].quantile(q)}
    quantiles.append(row)
```

```python
quantiled_data = pd.DataFrame(quantiles)
```

```python
quantiled_data
```

```python
quantiled_data.to_csv("./quantiled.csv")
```
