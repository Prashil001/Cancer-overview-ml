```python
from collections import defaultdict
from itertools import combinations
```

```python
def cooccurrence_matrix(window_size):
    vocab = set()
    cooccur_dict = defaultdict(int)

    corpus_updated = r"final_output.txt"
    with open(corpus_updated,'r') as f:
        words = f.read().split()
        vocab.update(words)

        for i in range(len(words) - window_size + 1):
            window = words[i:i+window_size]
            all_combinations = combinations(set(window),2)
            for w1,w2 in all_combinations:
                if w1!=w2:
                    pair = tuple(sorted([w1,w2]))
                    cooccur_dict[pair] += 1

    return cooccur_dict, sorted(vocab)
```

```python
cooccur_dict, vocab = cooccurrence_matrix(10)
print(f"Total unique word pairs: {len(cooccur_dict)}") 
```

```python
import numpy as np
import pandas as pd
from scipy.sparse import lil_matrix
```

```python
vocab_index = {word:idx for idx,word in enumerate(vocab)}
size = len(vocab)
matrix = lil_matrix((size, size), dtype=int)

for (w1,w2), count in cooccur_dict.items():
    i,j = vocab_index[w1], vocab_index[w2]
    matrix[i,j] = count
    matrix[j,i] = count

cooccur_df = pd.DataFrame(matrix, index=vocab, columns=vocab)
```
