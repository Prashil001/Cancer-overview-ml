```python
import os
import glob
import pandas as pd
import numpy as np
```

```python
gene_file_path = str(GENE_FILE)
documents_path = str(CLEANED_TEXT_DIR)
textfilesList = glob.glob(os.path.join(documents_path, "*.txt"))
```

```python
gene_file = pd.read_csv(gene_file_path)
gene_file = gene_file.drop(columns=["Unnamed: 0","#tax_id"])
```

```python
gene_file["Symbol"].isnull().sum()
```

```python
gene_file
```

```python
from collections import defaultdict
import re

gene_map = defaultdict(list)


gene_set = set(gene_file["Symbol"].str.lower())

for i, file in enumerate(textfilesList):
    with open(file, "r", encoding="utf-8") as f:
        doc_text = f.read().lower()
        
        words = re.findall(r'\b\w+\b', doc_text)
        doc_words = set(words)  
        matched_genes = gene_set.intersection(doc_words)

        for gene in matched_genes:
            gene_map[gene].append(i)
```
