```python
def extractUnimp(gram,n,docCount):
    gram_unimportant = pd.Series()
    junkgram = gram.loc[gram['documents']>docCount][f'{n}-gram']
    for term in junkgram:
        gram_unimportant = pd.concat([gram_unimportant,pd.Series(term.split(" "))])

    return pd.Series(gram_unimportant.unique())
```

```python
def cleanNgram(gram,n):
    keepList = []
    myQuad = gram
    for terms in myQuad[f"{n}-gram"]:
        tempTerms = pd.Series(terms.split(" "))
        valCounts = pd.Series(tempTerms.value_counts()).sort_values(ascending=False)
        uniqueTerms = len(valCounts)
        maxFreq = valCounts.iloc[0]
        if maxFreq >= 2:
            keepList.append(False)
            continue
        keepList.append(True)
    goodnGrams = myQuad.loc[keepList]
    badnGrams = myQuad.loc[[not bo for bo in keepList]]
    return goodnGrams
```
