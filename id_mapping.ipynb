```python
import numpy as np
import pandas as pd
import glob
import os
```

```python
testFile_dir = r"D:\ML Internship\test 2\cleaned text file"
uni = r"goodUni_cleaned.csv"
bi = r"goodbi_cleaned.csv"
```

```python
def extractGram(csv_file,n):
    gram = []
    file = pd.read_csv(csv_file)
    for term in file[f'{n}-gram']:
        gram.append(term)
    return gram
```

```python
uni_gram_li = extractGram(uni,1)
bi_gram_li = extractGram(bi,2)
```

```python
mapping = dict()
```

```python
for idx, filepath in enumerate(glob.glob(os.path.join(testFile_dir, "*.txt"))):
    with open(filepath, 'r', encoding='utf-8') as infile:
        text = infile.read().split()  
        for word in text:
            if word in uni_gram_li:
                if word not in mapping:
                    mapping[word] = []
                elif idx in mapping[word]:
                    continue
                mapping[word].append(idx)
```
